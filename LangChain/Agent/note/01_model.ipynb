### 2、langchain 1.0 提供了统一初始化模型的方法 init_chat_model

```python
# 底层还是调用的ChatDeepSeek()

from langchain.chat_models import init_chat_model

model = init_chat_model(
    model="deepseek-v4-flash", # 底层调的还是ChatDeepSeek()
    model_provider="deepseek",
    # api_key=DEEPSEEK_API_KEY,
    # base_url=DEEPSEEK_BASE_URL,
    temperature=0.7, # 模型温度，范围0-1，默认0.7
)
```

## 模型的调用
1、invoke(): 阻塞式，一次性返回完整的结果<br>
2、stream(): 阻塞式，流式输出，实时返回每个token<br>
3、batch(): 阻塞式，批量处理多个输入<br>
以及它们的异步版本<br>
4、ainvoke()<br>
5、astream()<br>
6、abatch()<br>

### 1、入参类型 (Message)<br>
LangChain的消息包含三个字段：Role，Content，Metadata

```python
# 1、入参字符串
response = model.invoke("巴威台风对苏州的影响如何")
```

```python
# 2、字典列表 (多轮对话场景)
messages = [
    {"role": "system", "content": "你是一个话少的助手"},
    {"role": "user", "content": "巴威台风对苏州的影响如何"},
    # {"role": "assistant", "content": "影响有限"},
    # {"role": "user", "content": "我刚才问了什么问题"},
]
response = model.invoke(messages)
```

```python
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage

# 3、消息对象列表
message_list = [
    SystemMessage(content="你是一个话少的助手"),
    HumanMessage(content="巴威台风对苏州的影响如何"),
    # AIMessage(content="好的，正在查找")
]
response3 = model.invoke(message_list)
```

### 2、invoke的返回值 (AIMessage)

```python
response = model.invoke([HumanMessage("1+1等于多少")])
```

```python
print(type(response)) # AIMessage
```

```python
from rich import print as rprint
rprint(response)
```

### 3、stream流式调用

```python
# stream返回的是迭代器
for chunk in model.stream("巴威对苏州的影响"):
    print(chunk.text, end="", flush=True)
```

### 4、batch批量调用

```python
messages = [
    "1+1=多少",
    "Reed这个名字有什么意义"
    "你的知识库到什么时候"
]
responses = model.batch(messages)
for response in responses:
    print("---------------")
    print(response)
```
