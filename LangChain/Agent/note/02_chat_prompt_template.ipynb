## 提示词模版

### 1、ChatPromptTemplate

```python
# 实例化
from langchain_core.prompts import ChatPromptTemplate
from rich import print as rprint

template = ChatPromptTemplate.from_messages(
    [
        ("system", "你是一个话少的助手"),
        ("human", "你叫什么名字"),
        ("ai", "我是{name}"),
        ("human", "{user_input}"),
    ]
)
```

#### 调用

```python
# invoke调用
result = template.invoke({"user_input": "你叫什么名字来着", "name": "Alice"})
rprint(result)
rprint(type(result))
```

```python
# format调用
result = template.format(name="Alice", user_input="你叫什么名字来着")
rprint(result)
rprint(type(result))
```
